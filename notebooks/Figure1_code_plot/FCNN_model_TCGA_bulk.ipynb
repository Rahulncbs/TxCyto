```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import seaborn as sns 
import random
import matplotlib.pyplot as plt
```

### tumor purity across TCGA sample

```python
#df_tcga_sample_purity=pd.read_csv("/data/kumarr17/figures_to_upload/supplementray_table/TCGA_mastercalls.abs_tables_purity.csv")
df_tcga_sample_purity=pd.read_csv("/data/kumarr17/TxCyto/results/TCGA_mastercalls.abs_tables_purity.csv")

print(df_tcga_sample_purity['purity'].median())
```

### Supplementray figure 1

```python
# Remove missing values
purity = df_tcga_sample_purity["purity"].dropna()

# Figure
fig, ax = plt.subplots(figsize=(5, 4))

sns.histplot(
    purity,
    bins=30,
    edgecolor="black",
    linewidth=0.6,
    ax=ax
)

# Median reference line
median_purity = purity.median()

ax.axvline(
    median_purity,
    linestyle="--",
    linewidth=1.5,
    label=f"Median = {median_purity:.2f}"
)

# Labels
ax.set_xlabel("Tumor purity", fontsize=12)
ax.set_ylabel("Number of samples", fontsize=12)

ax.tick_params(axis="both", labelsize=10)

# Clean manuscript style
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(
    frameon=False,
    fontsize=10
)

plt.tight_layout()

# Save
#plt.savefig(
#    "/data/kumarr17/figures_to_upload/suplementary_figures/TCGA_tumor_purity_histogram.svg",
#    format="svg",
#    bbox_inches="tight"
#)

#plt.savefig(
#    "/data/kumarr17/figures_to_upload/suplementary_figures/TCGA_tumor_purity_histogram.pdf",
#    format="pdf",
#    bbox_inches="tight"
#)

#plt.show()
```
